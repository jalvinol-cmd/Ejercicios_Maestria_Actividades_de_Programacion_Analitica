## Importación de librerías y preparación de los datos

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/plotly/datasets/master/supermarket_Sales.csv"

data = pd.read_csv(url)

# Renombrar algunas columnas
data = data.rename(columns={
    'Tax 5%': 'Tax',
    'Cost of goods sold': 'Cogs',
    'Gross margin percentage': 'Gross margin pct',
    'Customer stratification rating': 'Rating'
})

# Normalizar los nombres de las columnas
data.columns = (
    data.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

data.head()
```

Después de ejecutar el proceso, los nombres de las columnas quedan estandarizados en minúscula y sin espacios, facilitando su identificación y posterior procesamiento. Las columnas resultantes son: invoice_id, branch, city, customer_type, gender, product_line, unit_price, quantity, tax, total, date, time, payment, cogs, gross_margin_pct, gross_income y rating.

### Ejercicio 14: varias métricas por método de pago

**Nivel:** Avanzado

Determine para cada método de pago:

- Cuántas facturas registra.
- Cuántas unidades se vendieron.
- Cuánto ingreso generó.

Una sola agrupación debe entregar las tres métricas con nombres explícitos.

```python
resumen = (
    data
    .groupby('payment')
    .agg(
        facturas=('invoice_id', 'size'),
        unidades=('quantity', 'sum'),
        ingreso=('total', 'sum')
    )
    .reset_index()
)

print(resumen.round(2))
```

# **F. Integrar información**

### Ejercicio 15: agregar la zona de cada sucursal

**Nivel:** Intermedio

La tabla de ventas no indica en qué zona está cada sucursal. Esa información se encuentra en un catálogo independiente.

Integre la información conservando todas las ventas.

```python
sucursales = pd.DataFrame({
    'branch': ['A', 'B', 'C'],
    'zona': ['Centro', 'Norte', 'Sur']
})

sucursales
```

```python
resultado = data.merge(
    sucursales,
    on='branch',
    how='left'
)

print("Dimensiones del resultado:", resultado.shape)

resultado[
    ['branch', 'city', 'zona', 'total']
].head()
```

# ¿Qué condición deben cumplir dos tablas para poder integrarse?

En este caso, la columna común es branch, que permite identificar y relacionar la sucursal correspondiente en ambas tablas. De esta manera, la información puede combinarse correctamente tomando esta columna como punto de relación.

# **# Ejercicio 16: sucursal líder entre los clientes Member**

**Nivel:** Avanzado

Considere únicamente las ventas de clientes de tipo `Member`.

Determine:

1. Qué sucursal genera el mayor ingreso total.
2. Cuál es su ticket promedio.
3. Cómo se compara ese ticket promedio con el de las otras sucursales.

El ticket promedio se calcula como:

\[
\text{Ticket promedio} =
\frac{\text{Ingreso total}}{\text{Número de facturas}}
\]

```python
# 1. Filtrar las ventas de clientes Member
member = data[
    data['customer_type'] == 'Member'
]

# 2. Agrupar por sucursal y calcular
#    el número de facturas y el ingreso total
resumen = (
    member
    .groupby('branch')
    .agg(
        facturas=('invoice_id', 'size'),
        ingreso=('total', 'sum')
    )
    .reset_index()
)

# 3. Calcular el ticket promedio
resumen['ticket_promedio'] = (
    resumen['ingreso'] /
    resumen['facturas']
)

# 4. Ordenar las sucursales de mayor a menor ingreso
resumen = resumen.sort_values(
    'ingreso',
    ascending=False
)

# 5. Mostrar el resultado
print(resumen.round(2))
```

**Conclusión**

La sucursal C registra el mayor ingreso entre los clientes de tipo Member, con un total de 56.881,28 y un ticket promedio de 336,58.

Además, presenta un ticket promedio superior al de las sucursales A y B, lo que indica que, en promedio, los clientes Member realizan compras de mayor valor en la sucursal C.
